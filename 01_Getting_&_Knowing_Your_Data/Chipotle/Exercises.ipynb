### Step 10. For the most-ordered item, how many items were ordered?

```python
chickenBowlOrder = df[df['item_name'] == 'Chicken Bowl']
totalOrders = sum(chickenBowlOrder['quantity'])
totalOrders
```

### Step 11. What was the most ordered item in the choice_description column?

```python
groupByChoiceDesc = df.groupby('choice_description').sum()
groupByChoiceDesc = groupByChoiceDesc.sort_values('quantity', ascending=False)
groupByChoiceDesc.head(1)
```

### Step 12. How many items were orderd in total?

```python
groupByChoiceDesc['quantity'].sum()
```

### Step 13. Turn the item price into a float

#### Step 13.a. Check the item price type

```python
df.item_price.dtype
```

#### Step 13.b. Create a lambda function and change the type of item price

```python
dollarizer = lambda x: float(x[1:-1])
df.item_price = df.item_price.apply(dollarizer)
df.item_price.head(10)
```

#### Step 13.c. Check the item price type

```python
df.item_price.dtype
```

### Step 14. How much was the revenue for the period in the dataset?
